```python
import pandas as pd
```

```python
trips = pd.read_parquet('green_tripdata_2025-11.parquet')
zones = pd.read_csv('taxi_zone_lookup.csv')
```

```python
""" For the trips in November 2025 (lpep_pickup_datetime between '2025-11-01' and '2025-12-01', exclusive of the upper bound), 
how many trips had a trip_distance of LESS than or equal to 1 mile?
"""
OneMiletrips = len(trips.query('trip_distance <= 1'))
print(f'Answer 3, One mile trips count: {OneMiletrips}')
```

```python
""" Which was the pick up day with the longest trip distance?
Only consider trips with trip_distance less than 100 miles (to exclude data errors).
"""
longTrip = trips.query('trip_distance < 100').sort_values(by='trip_distance', ascending=False)
# longTrip.head(1)
distance = longTrip['trip_distance'].iloc[0]
answer4 = longTrip['lpep_pickup_datetime'].iloc[0]
print(f'Answer 4, pickup day with longest trip distance: "{answer4}" with a {distance} Miles of distance')
```

```python

""" Which was the pickup zone with the largest total_amount (sum of all trips) on November 18th, 2025?
"""
joindata = trips.merge(zones, left_on='PULocationID' , right_on='LocationID', how='left')
largestamount = joindata.groupby('Zone')['total_amount'].sum().reset_index().sort_values(by='total_amount', ascending=False)
# largestamount.head(1)
answer5 = largestamount['Zone'].iloc[0]
amount = largestamount['total_amount'].iloc[0]
print(f'Answer 5, pickup day with longest tripdistance: {answer5} , with sum  of ${amount}')
```

```python
""" For the passengers picked up in the zone named "East Harlem North" in November 2025, 
which was the drop off zone that had the largest tip?

Note: it's tip , not trip. We need the name of the zone, not the ID.
"""
bigtip = joindata.query('PULocationID == 74').sort_values(by='tip_amount', ascending=False)
tipamount = bigtip['tip_amount'].iloc[0]
# bigtip[['PULocationID','tip_amount','DOLocationID']].head(1)
### Get DOLocationID
dropID = bigtip['DOLocationID'].iloc[0]
### Get Zone
zone = zones.query('LocationID == @dropID')
### Get column value
dropOffzone = zone['Zone'].iloc[0]
print(f'Dropoff zone with the lasrgest tip: "{dropOffzone}" with tip amount: ${tipamount}')
```
