```python
import pandas as pd
import click
from sqlalchemy import create_engine
import pyarrow.parquet as pq
from tqdm.auto import tqdm
```

```python
df = pd.read_parquet('green_tripdata_2025-11.parquet')
```

```python
engine = create_engine('postgresql://root:root@localhost:5432/ny_taxi')
```

```python
#Print schema of parquet data:
# print(pd.io.sql.get_schema(df, name='green_tripdata_2025-11', con=engine))
df.shape
```

```python
# Create only schema head(0) for green_tripdata_2025-11 data ****
df.head(0).to_sql(name='green_tripdata_2025-11', con=engine, if_exists='replace')
```

```python
# Open the parquet file with pyarrow.parquet
parquet_file = pq.ParquetFile('green_tripdata_2025-11.parquet')

# Iterate through the file in chunks (batches)
for batch in parquet_file.iter_batches(batch_size=10000):
    df_chunk = batch.to_pandas()
    df_chunk.to_sql(name='green_tripdata_2025-11', con=engine, if_exists='append')
    # Process your chunk here
    print(f"Loaded chunk with {len(df_chunk)} rows")
```
